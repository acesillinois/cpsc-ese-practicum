```python
# 3 features, 2 informative, 1 redundant - 3D plot

# Import necessary libraries
import numpy as np
import plotly.graph_objects as go
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

# 1. Generate synthetic binary classification data
n_features = 3  # We will use 3 features for the 3D plot
X, y = make_classification(n_samples=1000,
                           n_features=n_features,
                           n_informative=2,
                           n_redundant=1,
                           n_repeated=0,
                           random_state=42)

# Add intercept term to X (i.e., a column of ones for the bias term)
X = np.hstack((np.ones((X.shape[0], 1)), X))

# 2. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Define logistic regression functions
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_cost(X, y, beta):
    m = len(y)
    h = sigmoid(X @ beta)
    cost = -(1/m) * (y.T @ np.log(h) + (1 - y).T @ np.log(1 - h))
    return cost

def compute_gradient(X, y, beta):
    m = len(y)
    h = sigmoid(X @ beta)
    gradient = (1/m) * X.T @ (h - y)
    return gradient

def gradient_descent(X, y, beta, learning_rate, iterations):
    costs = []
    for i in range(iterations):
        gradient = compute_gradient(X, y, beta)
        beta = beta - learning_rate * gradient
        cost = compute_cost(X, y, beta)
        costs.append(cost)
    return beta, costs

# 4. Initialize parameters (weights)
beta = np.zeros(X_train.shape[1])  # Initialize beta with zeros; X_train has 4 columns, so beta has 4 elements

# 5. Perform gradient descent
learning_rate = 0.01
iterations = 1000
beta_opt, costs = gradient_descent(X_train, y_train, beta, learning_rate, iterations)

# 6. Define prediction function
def predict(X, beta, threshold=0.5):
    probs = sigmoid(X @ beta)
    return (probs >= threshold).astype(int)

# 7. Plot 3D decision boundary with Plotly
def plot_3d_decision_boundary(X, y, beta):
    x1_min, x1_max = X[:, 1].min(), X[:, 1].max()
    x2_min, x2_max = X[:, 2].min(), X[:, 2].max()
    x3_min, x3_max = X[:, 3].min(), X[:, 3].max()

    # Create a 3D grid for plotting
    xx1, xx2, xx3 = np.meshgrid(np.linspace(x1_min, x1_max, 50),
                                np.linspace(x2_min, x2_max, 50),
                                np.linspace(x3_min, x3_max, 50))

    # Prepare the grid for prediction (ensure matching dimensionality)
    grid = np.c_[np.ones((xx1.size, 1)), xx1.ravel(), xx2.ravel(), xx3.ravel()]  # Add intercept term

    # Predict probabilities using the logistic regression model
    probs = sigmoid(grid @ beta)  # Compute the probabilities

    # Reshape probabilities to match the grid shape for plotting
    probs = probs.reshape(xx1.shape)

    # Plot 3D decision boundary and data points using Plotly
    fig = go.Figure()

    # Add 3D isosurface plot for the decision boundary
    fig.add_trace(go.Isosurface(
        x=xx1.ravel(),
        y=xx2.ravel(),
        z=xx3.ravel(),
        value=probs.ravel(),
        isomin=0.5,  # Decision boundary at probability 0.5
        isomax=0.5,
        surface_count=1,
        colorscale='RdBu',
        opacity=0.6
    ))

    # Add scatter plot for the actual data points
    fig.add_trace(go.Scatter3d(
        x=X[:, 1],
        y=X[:, 2],
        z=X[:, 3],
        mode='markers',
        marker=dict(color=y, colorscale='RdBu', line=dict(width=1, color='White')),
        showlegend=False
    ))

    # Customize plot
    fig.update_layout(
        title="3D Decision Boundary with Logistic Regression",
        scene=dict(
            xaxis_title="Feature 1",
            yaxis_title="Feature 2",
            zaxis_title="Feature 3"
        ),
        width=800, height=800
    )

    fig.show()

# Call the plot function with test data
plot_3d_decision_boundary(X_test, y_test, beta_opt)
```

```python
# Import necessary libraries
import numpy as np
import plotly.graph_objects as go
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

# 1. Generate synthetic binary classification data with 2 informative features
n_features = 2  # Two features (both informative) for binary classification
X, y = make_classification(n_samples=1000,
                           n_features=n_features,
                           n_informative=2,
                           n_redundant=0,
                           n_classes=2,                # Binary classification
                           random_state=42)

# Add intercept term to X (i.e., a column of ones for the bias term)
X = np.hstack((np.ones((X.shape[0], 1)), X))  # X now has 3 columns (1 for intercept, 2 for features)

# 2. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Define logistic regression functions
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_cost(X, y, beta):
    m = len(y)
    h = sigmoid(X @ beta)
    cost = -(1/m) * (y.T @ np.log(h) + (1 - y).T @ np.log(1 - h))
    return cost

def compute_gradient(X, y, beta):
    m = len(y)
    h = sigmoid(X @ beta)
    gradient = (1/m) * X.T @ (h - y)
    return gradient

def gradient_descent(X, y, beta, learning_rate, iterations):
    costs = []
    for i in range(iterations):
        gradient = compute_gradient(X, y, beta)
        beta = beta - learning_rate * gradient
        cost = compute_cost(X, y, beta)
        costs.append(cost)
    return beta, costs

# 4. Initialize parameters (weights)
beta = np.zeros(X_train.shape[1])  # Initialize beta with zeros; X_train has 3 columns, so beta has 3 elements

# 5. Perform gradient descent
learning_rate = 0.01
iterations = 1000
beta_opt, costs = gradient_descent(X_train, y_train, beta, learning_rate, iterations)

# 6. Define prediction function
def predict(X, beta, threshold=0.5):
    probs = sigmoid(X @ beta)
    return (probs >= threshold).astype(int)

# 7. Plot decision boundary in 2D with Plotly
def plot_2d_decision_boundary(X, y, beta):
    x1_min, x1_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    x2_min, x2_max = X[:, 2].min() - 1, X[:, 2].max() + 1

    # Create a mesh grid for plotting
    xx1, xx2 = np.meshgrid(np.linspace(x1_min, x1_max, 100), np.linspace(x2_min, x2_max, 100))

    # Prepare the grid for prediction (ensure matching dimensionality)
    grid = np.c_[np.ones((xx1.size, 1)), xx1.ravel(), xx2.ravel()]  # Add intercept term

    # Predict probabilities using the logistic regression model
    probs = sigmoid(grid @ beta)  # Compute the probabilities

    # Reshape probabilities to match the grid shape for plotting
    probs = probs.reshape(xx1.shape)

    # Plot decision boundary and data points using Plotly
    fig = go.Figure()

    # Add contour plot for decision boundary
    fig.add_trace(go.Contour(
        z=probs,
        x=np.linspace(x1_min, x1_max, 100),
        y=np.linspace(x2_min, x2_max, 100),
        colorscale='RdBu',
        opacity=0.8,
        contours=dict(start=0.0, end=1.0, size=0.5),  # Contour at probability 0.5
        showscale=False
    ))

    # Add scatter plot for the actual data points
    fig.add_trace(go.Scatter(
        x=X[:, 1],
        y=X[:, 2],
        mode='markers',
        marker=dict(color=y, colorscale='RdBu', line=dict(width=1, color='Black')),
        showlegend=False
    ))

    # Customize plot
    fig.update_layout(
        title="2D Decision Boundary with Logistic Regression",
        xaxis_title="Feature 1",
        yaxis_title="Feature 2",
        xaxis=dict(range=[x1_min, x1_max]),
        yaxis=dict(range=[x2_min, x2_max]),
        width=800, height=600
    )

    fig.show()

# Call the plot function with test data
plot_2d_decision_boundary(X_test, y_test, beta_opt)
```
